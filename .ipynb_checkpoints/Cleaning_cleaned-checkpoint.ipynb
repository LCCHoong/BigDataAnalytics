```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
from pyspark.sql.functions import col, when, sum, mean, min, max, count, lit
from pyspark.sql.types import StringType
```

```python
spark = SparkSession.builder.getOrCreate()
```

```python
martBranches = spark.read.csv("mart_branches.csv",       header=True)
martCust     = spark.read.csv("mart_customers.csv",      header=True)
martProd     = spark.read.csv("mart_products.csv",       header=True)
martProm     = spark.read.csv("mart_promotions.csv",     header=True)
martSales    = spark.read.csv("mart_sales_master.csv",   header=True)
```

```python
# ─── martBranches ───────────────────────────────────────────────────────────
# Schema is all-string which is fine for a reference table.
# Only action needed: drop duplicate branch_code rows.

martBranches = martBranches.dropDuplicates(["branch_code"])
print("martBranches rows:", martBranches.count())
martBranches.show()
```

```python
#  martProd 
# 1. Remove the bad 'Potato Chips Barbeque' record.
# 2. Cast price columns to float.
# 3. Confirm no nulls / no duplicates.

martProd = martProd.filter(F.col("product_name") != "Potato Chips Barbeque")

martProd = (
    martProd
    .withColumn("std_price",  col("std_price").cast("float"))
    .withColumn("cost_price", col("cost_price").cast("float"))
)

total_rows_prod = martProd.count()
distinct_prod   = martProd.distinct().count()
print(f"martProd rows: {total_rows_prod}  |  duplicates: {total_rows_prod - distinct_prod}")

# Null check
martProd.select([
    F.count(F.when(F.col(c).isNull(), c)).alias(c)
    for c in martProd.columns
]).show()

martProd.show(truncate=False)
```

```python
# martProm 
# 1. Null check.
# 2. Duplicate check.
# 3. Remove 'BADPROMO' reference record.

martProm.select([
    F.count(F.when(F.col(c).isNull(), c)).alias(c)
    for c in martProm.columns
]).show()

print("martProm duplicates:", martProm.count() - martProm.distinct().count())

martProm.show()
```

```python
# martCust

total_rows_cust = martCust.count()
print("martCust rows:", total_rows_cust)

# Cast types
martCust = (
    martCust
    .withColumn("age", col("age").cast("integer"))
    .withColumn("member_since", col("member_since").cast("date"))
)

martCust.printSchema()
```

```python
distinct_cust = martCust.distinct().count()
print(f"martCust duplicate rows: {total_rows_cust - distinct_cust}")
```

```python
# Age: flag and null out impossible values (>= 150)
print("Age distribution before fix:")
martCust.select(
    F.skewness("age").alias("age_skew")
).show()
martCust.select(
    F.percentile_approx("age", [0, 0.25, 0.5, 0.75, 1]).alias("quartiles")
).show(truncate=False)

print("Rows with age >= 150:", martCust.filter("age >= 150").count())

martCust = martCust.withColumn(
    "age",
    F.when(F.col("age") >= 150, None).otherwise(F.col("age"))
)

print("Age distribution after fix:")
martCust.select(F.skewness("age").alias("age_skew")).show()
martCust.select(
    F.percentile_approx("age", [0, 0.25, 0.5, 0.75, 1]).alias("quartiles")
).show(truncate=False)
```

```python
# member_since: null out dates outside the programme window (2021-01-01 to today)
print("member_since before 2021-01-01:",
      martCust.filter(F.col("member_since") < F.lit("2021-01-01")).count())
print("member_since after 2026-06-01:",
      martCust.filter(F.col("member_since") > F.lit("2026-06-01")).count())

martCust = martCust.withColumn(
    "member_since",
    F.when(
        (F.col("member_since") < F.lit("2021-01-01")) |
        (F.col("member_since") > F.lit("2026-06-01")),
        None
    ).otherwise(F.col("member_since"))
)

print("member_since nulls after bound fix:",
      martCust.filter(F.col("member_since").isNull()).count())
```

```python
# Null counts before fill
print("Null counts per column:")
martCust.select([
    F.count(F.when(F.col(c).isNull(), c)).alias(c)
    for c in martCust.columns
]).show()

# gender  → mode
mode_gender = (
    martCust.groupBy("gender").count()
    .orderBy("count", ascending=False)
    .first()[0]
)
martCust = martCust.fillna({"gender": mode_gender})

# age     → mean
mean_age = martCust.select(mean("age")).first()[0]
martCust = martCust.fillna({"age": mean_age})

# member_tier → 'Unmembered'
martCust = martCust.fillna({"member_tier": "Unmembered"})

# Verify
print("Null counts after fill:")
martCust.select([
    F.count(F.when(F.col(c).isNull(), c)).alias(c)
    for c in martCust.columns
]).show()
```

```python
# customer_id uniqueness
dup_cust = martCust.groupBy("customer_id").count().filter("count > 1")
print("Duplicate customer_ids:", dup_cust.count())

martCust.groupBy("gender").count().show()
print("martCust rows:", martCust.count())
```

```python
#martSales 
total_rows_sales = martSales.count()
print("martSales rows:", total_rows_sales)
martSales.show(truncate = False)
martSales.printSchema()
```

```python
# Parse the mixed-format timestamp column into a single standard column.
# Formats observed in data:
#   dd/MM/yyyy HH:mm
#   yyyy-MM-dd HH:mm:ss
#   MMM dd yyyy HH:mm
#   MM-dd-yyyy HH:mm

print("txn_raw nulls before parse:",
      martSales.filter(F.col("txn_raw").isNull()).count())

martSales = martSales.withColumn(
    "txn_cooked",
    F.coalesce(
        F.to_timestamp("txn_raw", "dd/MM/yyyy HH:mm"),
        F.to_timestamp("txn_raw", "yyyy-MM-dd HH:mm:ss"),
        F.to_timestamp("txn_raw", "MMM dd yyyy HH:mm"),
        F.to_timestamp("txn_raw", "MM-dd-yyyy HH:mm"),
    )
)

# Show any values that still couldn't be parsed
unparsed = (
    martSales.filter(F.col("txn_cooked").isNull())
    .groupBy("txn_raw").count()
)
print("Unparsed txn_raw values:")
unparsed.show(100, False)
print("Unparsed count:", martSales.filter(F.col("txn_cooked").isNull()).count())

martSales = (
    martSales
    .drop("txn_raw")
    .withColumnRenamed("txn_cooked", "transaction_datetime")
)

martSales.groupBy("transaction_datetime").count().orderBy("count", ascending=False).show (truncate=False)
```

```python
null_pct_txn = (
    martSales.filter(F.col("transaction_datetime").isNull()).count()
    / total_rows_sales * 100
)
print("null pct " , null_pct_txn)

# Fill with mode 
mode_txn = (
    martSales.filter(F.col("transaction_datetime").isNotNull())
    .groupBy("transaction_datetime").count()
    .orderBy("count", ascending=False)
    .first()[0]
)
martSales = martSales.withColumn(
    "transaction_datetime",
    F.when(F.col("transaction_datetime").isNull(), F.lit(mode_txn))
     .otherwise(F.col("transaction_datetime"))
)
print("transaction_datetime nulls after fill:",
      martSales.filter(F.col("transaction_datetime").isNull()).count())
```

```python
# Standardise branch names to branch codes; unknown values → null → filled with mode.

BRANCH_MAP = {
    # Petaling Jaya
    "Petaling Jaya": "PJ", "Petaling Jayaa": "PJ", "PetalingJaya": "PJ",
    "PetalingJya": "PJ", "PETALING JAYA": "PJ", "petaling jaya": "PJ", "P.J.": "PJ",
    # Subang
    "Subang": "SB", "SubangJaya": "SB", "Subang Jaya": "SB",
    "subang jaya": "SB", "SUBANG JAYA": "SB", "Subng Jaya": "SB",
    # Kuala Lumpur
    "Kuala Lumpur": "KL", "kuala lumpur": "KL", "KUALA LUMPUR": "KL",
    "KualaLumpur": "KL", "Kuala Lumper": "KL", "K.L.": "KL",
    # Johor Bahru
    "Johor Bahru": "JB", "JOHOR BAHRU": "JB", "johor bahru": "JB", "JohorBahru": "JB",
    # Others
    "Penang": "PG", "penang": "PG", "PENANG": "PG",
    "Ipoh": "IP", "IPOH": "IP", "ipoh": "IP",
    "Cheras": "CH", "CHERAS": "CH", "cheras": "CH",
    # Invalid
    "ERROR": None, "UNKNOWN": None,
}

martSales = martSales.replace(BRANCH_MAP, subset=["branch"])

null_pct_br = (
    martSales.filter(F.col("branch").isNull()).count()
    / total_rows_sales * 100
)
print(f"branch null % after map: {null_pct_br:.2f}%")

mode_branch = (
    martSales.groupBy("branch").count()
    .orderBy("count", ascending=False)
    .first()[0]
)
martSales = martSales.fillna({"branch": mode_branch})

print("branch nulls after fill:",
      martSales.filter(F.col("branch").isNull()).count())
martSales.groupBy("branch").count().show(100, False)
```

```python
# qty: cast to int, null out negatives + sentinel 9999+, cap at 99th percentile,
# fill remaining nulls with median.

martSales = martSales.withColumn("qty", col("qty").cast("integer"))

# Null out impossible values
martSales = martSales.withColumn(
    "qty",
    F.when((F.col("qty") < 0) | (F.col("qty") >= 9999), None)
     .otherwise(F.col("qty"))
)

# Cap outliers at 99th percentile
upper_qty = martSales.approxQuantile("qty", [0.99], 0.01)[0]
martSales = martSales.withColumn(
    "qty",
    F.when(F.col("qty") > upper_qty, upper_qty).otherwise(F.col("qty"))
)

# Fill nulls with median
median_qty = martSales.approxQuantile("qty", [0.5], 0.01)[0]
martSales = martSales.fillna({"qty": int(median_qty)})

print("qty nulls after fill:", martSales.filter(F.col("qty").isNull()).count())
martSales.select(
    F.skewness("qty").alias("qty_skew"),
    F.percentile_approx("qty", [0, 0.25, 0.5, 0.75, 1]).alias("quartiles")
).show(truncate=False)
```

```python
# unit_price: strip 'RM' prefix, map 'free' → 0, cast to float.

martSales = martSales.withColumn(
    "unit_price",
    when(col("unit_price").startswith("RM"),
         col("unit_price").substr(3, 100).cast("float"))
    .when(col("unit_price") == "free", lit(0).cast("float"))
    .otherwise(col("unit_price").cast("float"))
)

print("unit_price nulls:", martSales.filter(F.col("unit_price").isNull()).count())

martSales.filter(F.col("unit_price").isNull()).show()

##
price_lookup = martProd.select(
    "product_id",
    col("std_price").alias("correct_price"),
    col("cost_price").alias("correct_cost")
)

# Extract discount % from description e.g. "Member 8%" → 0.08
promo_lookup = martProm.select(
    "promo_code",
    (F.regexp_extract(F.col("description"), r"(\d+)%", 1).cast("float") / 100)
    .alias("discount_rate")
)

martSales = (
    martSales
    .join(price_lookup, on="product_id", how="left")
    .join(promo_lookup, on="promo_code", how="left")
    .withColumn("unit_price",
        F.when(
            F.col("discount_rate").isNotNull() & (F.col("discount_rate") > 0),
            F.col("correct_price") * (1 - F.col("discount_rate"))
        ).otherwise(F.col("correct_price"))
    )

        
    .withColumn("cost_price", F.col("correct_cost"))
    .drop("correct_price", "correct_cost", "discount_rate")


)

print("unit_price nulls:", martSales.filter(F.col("unit_price").isNull()).count())
martSales = martSales.withColumn("unit_price", F.round(F.col("unit_price"), 2))
martSales.select("product_id", "promo_code", "unit_price", "cost_price").show(20)
```

```python
# Overwrite product_name using martProd as the authoritative source.
# Drop any sales rows whose product_id has no match (truly orphaned records).
lookup = martProd.select("product_id", col("product_name").alias("prod_name_clean"))

martSales = (
    martSales
    .join(lookup, on="product_id", how="left")
    .withColumn(
        "product_name",
        F.when(F.col("prod_name_clean").isNotNull(), F.col("prod_name_clean"))
         .otherwise(F.col("product_name"))  # keep original if lookup missed
    )
    .drop("prod_name_clean")
)

# Drop rows with no product_id (cannot be linked to any product)
martSales = martSales.filter(F.col("product_id").isNotNull())
print("martSales rows after product_id null drop:", martSales.count())

# Check remaining nulls
print("product_name nulls after lookup:", martSales.filter(F.col("product_name").isNull()).count())

martSales.filter(F.col("product_name").isNull()).show()
```

```python
distinct_sales = martSales.distinct().count()
print(f"Duplicate rows before dedup: {martSales.count() - distinct_sales}")
martSales = martSales.distinct()
print(f"martSales rows after dedup: {martSales.count()}")

valid_product  = martProd.select("product_id")
valid_customer = martCust.select("customer_id")
valid_branch   = martBranches.select(col("branch_code").alias("branch"))
valid_promo    = martProm.select("promo_code")

martSales = (
    martSales
    .join(valid_product,  on="product_id",  how="inner")
    .join(valid_customer, on="customer_id", how="inner")
    .join(valid_branch,   on="branch",      how="inner")
    .join(valid_promo,    on="promo_code",  how="inner")
)

print("martSales rows after referential integrity drop:", martSales.count())
```

```python
# Check for dangling foreign keys across all fact → dimension joins.

missing_cust = martSales.join(
    martCust.select("customer_id"), on="customer_id", how="left_anti"
)
missing_prod = martSales.join(
    martProd.select("product_id"), on="product_id", how="left_anti"
)
missing_branch = martSales.join(
    martBranches.select(col("branch_code").alias("branch")),
    on="branch", how="left_anti"
)
missing_promo = martSales.join(
    martProm.select("promo_code"), on="promo_code", how="left_anti"
)

print("Sales rows with unknown customer_id:", missing_cust.count())
print("Sales rows with unknown product_id: ", missing_prod.count())
print("Sales rows with unknown branch:     ", missing_branch.count())
print("Sales rows with unknown promo_code: ", missing_promo.count())
```

```python
# Summary: row count + null counts for every cleaned dataframe.

dfs = {
    "martBranches": martBranches,
    "martCust":     martCust,
    "martProd":     martProd,
    "martProm":     martProm,
    "martSales":    martSales,
}

for name, df in dfs.items():
    row_count   = df.count()
    null_counts = df.select([
        F.count(F.when(F.col(c).isNull(), c)).alias(c)
        for c in df.columns
    ]).collect()[0]
    print(f"  {name}  |  rows: {row_count}")
    print(f"  nulls: {null_counts.asDict()}")
```
